```python
from PIL import Image

def resize_img(img_path, output_path, size):
    img = Image.open(r"C:\Users\Student\Desktop\research papers\Cat.jpg")
    resized_img = img.resize(size)
    output_path = resized_img.save(r"C:\Users\Student\Desktop\research papers\resized_Cat.jpg")
    print("Image resized Successfully to ", size)

def crop_img(img_path, output_path, crop_box):
    img = Image.open(r"C:\Users\Student\Desktop\research papers\Cat.jpg")
    cropped_img = img.crop(crop_box)
    output_path = cropped_img.save(r"C:\Users\Student\Desktop\research papers\Cropped_Cat.jpg")
    print("Image Cropped Successfully to ", crop_box)

def flip_img(img_path, output_path, direction):
    img = Image.open(r"C:\Users\Student\Desktop\research papers\Cat.jpg")
    if direction == 'Horizontal':
        flip_img = img.transpose(Image.FLIP_LEFT_RIGHT)
        output_path = flip_img.save(r"C:\Users\Student\Desktop\research papers\Flipped_H_Cat.jpg")
    else:
        flip_img = img.transpose(Image.FLIP_TOP_BOTTOM)
        output_path = flip_img.save(r"C:\Users\Student\Desktop\research papers\Flipped_V_Cat.jpg")

    print("Image Flipped Successfully to ", direction)

def rotate_img(img_path, output_path, angle):
    img = Image.open(r"C:\Users\Student\Desktop\research papers\Cat.jpg")
    rotateded_img = img.rotate(angle)
    output_path = rotateded_img.save(r"C:\Users\Student\Desktop\research papers\Rotated_Cat.jpg")
    print("Image Rotated Successfully to ", angle)
    

img = Image.open(r"C:\Users\Student\Desktop\research papers\Cat.jpg")
resize_img(img, (r'C:\Users\Student\Desktop\research papers\resized_Cat.jpg'), (500, 500))
crop_img(img, (r'C:\Users\Student\Desktop\research papers\Cropped_Cat.jpg'), (500,600,700,900))
flip_img(img, (r'C:\Users\Student\Desktop\research papers\Flipped_Cat.jpg'), 'Vertical')
flip_img(img, (r'C:\Users\Student\Desktop\research papers\Flipped_Cat.jpg'), 'Horizontal')
rotate_img(img, (r'C:\Users\Student\Desktop\research papers\Rotated_Cat.jpg'), 270)
```
