```python
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np

img = Image.open(r"C:\Users\Student\Desktop\research papers\Cat.jpg").convert("L")
img = np.array(img)

h, w = img.shape
image = img.tolist()

def convolution(img, filter):
    h, w = img.shape
    fh, fw = filter.shape
    output = np.zeros((h, w))
    for i in range(1, h-1):
        for j in range(1, w-1):
            s = 0
            for m in range(fh):
                for n in range(fw):
                    s += img[i+m-1][j+n-1] * filter[m][n]
            output[i][j] = s
    return output

def Robert_filter(img):
    robert = np.array([[-1, 0],[ 0, 1]])
    r = convolution(img, robert)
    return r

def sobel_filter(img):
    sobel = np.array([[-1, -2, -1],[ 0,  0,  0],[ 1,  2,  1]])
    s = convolution(img, sobel)
    return s

def Prewitt_filter(img):
    prewitt = np.array([[-1, -1, -1],[ 0,  0,  0], [ 1,  1,  1]])
    p = convolution(img, prewitt)
    return p

def Mean_filter(img):
    mean = np.array([[1/9, 1/9, 1/9],[1/9, 1/9, 1/9],[1/9, 1/9, 1/9]])
    m = convolution(img, mean)
    return m

def Gaussian_Filter(img):
    gaussian = np.array([[1/16, 2/16, 1/16],[2/16, 4/16, 2/16],[1/16, 2/16, 1/16]])
    g = convolution(img, gaussian)
    return g

def Median_filter(img):
    print()

while True:
    print("Filtering Technique")
    print("1. Display Original Image")
    print("2. Robert Filter")
    print("3. Sobel Filter")
    print("4. Prewitt Filter")
    print("5. Mean Filter")
    print("6. Gaussian Filter")
    print("7. Median Filter")
    print("8. Max Filter")
    print("9. Min Filter")
    print("0. Exit")
    choice = int(input("Enter your choice: "))
    if choice == 1:
        print("Original Image:")
        plt.imshow(img, cmap='gray')
        plt.title("Original Image")
        plt.axis("off")
        plt.show()

    elif choice == 2:
        result = Robert_filter(img)
        print("Robert Filter Output:")
        plt.imshow(result, cmap='gray')
        plt.title("Robert Filter")
        plt.axis("off")
        plt.show()
        
    elif choice == 3:
        result = sobel_filter(img)
        print("Sobel Filter Output:")
        plt.imshow(result, cmap='gray')
        plt.title("Sobel Filter")
        plt.axis("off")
        plt.show()

    elif choice == 4:
        result = Prewitt_filter(img)
        print("Prewitt Filter Output:")
        plt.imshow(result, cmap='gray')
        plt.title("Prewitt Filter")
        plt.axis("off")
        plt.show()

    elif choice == 5:
        result = Mean_filter(img)
        print("Mean Filter Output:")
        plt.imshow(result, cmap='gray')
        plt.title("Mean Filter")
        plt.axis("off")
        plt.show()

    elif choice == 6:
        result = Gaussian_Filter(img)
        print("Gaussian Filter Output:")
        plt.imshow(result, cmap='gray')
        plt.title("Gaussian Filter")
        plt.axis("off")
        plt.show()

    elif choice == 7:
        result = Gaussian_Filter(img)
        print("Median Filter Output:")
        plt.imshow(result, cmap='gray')
        plt.title("Median Filter")
        plt.axis("off")
        plt.show()

    elif choice == 8:
        result = Gaussian_Filter(img)
        print("Max Filter Output:")
        plt.imshow(result, cmap='gray')
        plt.title("Max Filter")
        plt.axis("off")
        plt.show()

    elif choice == 9:
        result = Gaussian_Filter(img)
        print("Min Filter Output:")
        plt.imshow(result, cmap='gray')
        plt.title("Min Filter")
        plt.axis("off")
        plt.show()
        
    elif choice == 0:
        print("Program Ended...Exiting the program...^_^")

        break
    else:
        print("Invalid Choice!!!! Please enter a number between 1 and 7.")
```
