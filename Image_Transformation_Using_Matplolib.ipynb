```python
from PIL import Image
import matplotlib.pyplot as plt

img = Image.open(r"C:\Users\Student\Desktop\research papers\Cat.jpg")
resized = img.resize((500,500))
cropped = img.crop((50,60,200,300))
H_Flipped = img.transpose(Image.FLIP_LEFT_RIGHT)
V_Flipped = img.transpose(Image.FLIP_TOP_BOTTOM)
rotated = img.rotate(270)

img_Trans = [img, resized, cropped, H_Flipped, V_Flipped, rotated ]
title = ["Original", "Resized", "Cropped", "Horizontal Flipped","Vertical Flipped","Rotated" ]

plt.figure(figsize=(14,8))

for idx in range(6):
    plt.subplot(2,3,idx + 1) 
    plt.imshow(img_Trans[idx])
    plt.title(title[idx])
    plt.axis('off')

plt.tight_layout()
plt.show()
```
