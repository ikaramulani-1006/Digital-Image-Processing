```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


img = cv2.imread(r'C:\Users\Student\Desktop\research papers\Cat.jpg')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img)
plt.title("RGB Image")
plt.axis("off")
plt.show()

R = img[:, :, 0]
G = img[:, :, 1]
B = img[:, :, 2]

gray = (0.299 * R + 0.587 * G + 0.114 * B)
plt.imshow(gray, cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")
plt.show()


R = R/255
G = G/255
B = B/255
C_max = np.max([R, G, B], axis=0)
C_min = np.min([R, G, B], axis=0)
C_Delta = C_max - C_min


if np.all(C_max == C_min):
    h = 0
elif np.all(C_max == R):
    h = (60 * ((G - B/C_Delta)+360))%360
elif np.all(C_max == G):
    h = (60 * ((B - R/C_Delta)+120))%360
elif np.all(C_max == B):
    h = (60 * ((R - G/C_Delta)+240))%360
else:
    h = 0
if np.all(C_max != 0):
    s = (C_Delta/C_max)*100
else:
    s = 0

h = np.zeros_like(C_max)
s = np.zeros_like(C_max)
v = C_max*100

hsv_image = np.stack([h, s, v], axis=-1)
hsv_image = hsv_image.astype(np.uint8)
plt.imshow(hsv_image)
plt.title("HSV Image")
plt.axis("off")
plt.show()
```
